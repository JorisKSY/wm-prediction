# SoccerData / FBref Data Inventory

Goal: Explore `soccerdata` as a possible source for football match data.

This notebook tests a small FBref World Cup schedule query.

No cleaning, feature engineering, SQL import, or modeling in this notebook.

```python
from pathlib import Path
from importlib.metadata import version
import json

import pandas as pd
import soccerdata as sd

RAW_DIR = Path("../data/raw/soccerdata/fbref_world_cup")
RAW_DIR.mkdir(parents=True, exist_ok=True)

print("soccerdata version:", version("soccerdata"))
RAW_DIR
```

```python
available_fbref_leagues = sd.FBref.available_leagues()

world_cup_leagues = [
    league for league in available_fbref_leagues
    if "World Cup" in league
]

world_cup_leagues
```

```python
# Small test query: FBref World Cup 2022 schedule
# This is only a data inventory test, not final project data.

fbref = sd.FBref(
    leagues="INT-World Cup",
    seasons=2022,
    data_dir=RAW_DIR / "cache"
)
```

```python
schedule = fbref.read_schedule()

schedule.head()
```

```python
snapshot_name = "fbref_int_world_cup_2022_schedule_raw.csv"
csv_path = RAW_DIR / snapshot_name

schedule.reset_index().to_csv(csv_path, index=False)

csv_path
```

```python
metadata = {
    "provider": "FBref via soccerdata",
    "package": "soccerdata",
    "package_version": version("soccerdata"),
    "source_type": "web scraper wrapper",
    "soccerdata_class": "FBref",
    "leagues": "INT-World Cup",
    "seasons": 2022,
    "method": "read_schedule",
    "local_csv_file": snapshot_name,
    "local_cache_dir": str(RAW_DIR / "cache"),
    "notes": [
        "Local raw snapshot for data inventory.",
        "Not committed to Git.",
        "No cleaning, feature engineering, SQL import, or modeling performed.",
        "Data source depends on FBref website structure and may break if the website changes."
    ],
}

metadata_path = RAW_DIR / "fbref_int_world_cup_2022_schedule_metadata.json"

with open(metadata_path, "w") as f:
    json.dump(metadata, f, indent=2)

metadata_path
```
